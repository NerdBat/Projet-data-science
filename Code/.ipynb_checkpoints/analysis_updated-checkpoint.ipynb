# Analyse & comparaison des modèles — Churn Prediction

Ce notebook sert à **comparer les modèles** de classification du churn:

- mêmes **10 variables** que `train.py`, `api.py` et `app.py`,
- **cross-validation stratifiée**,
- **optimisation du seuil** avec le **F2-score**,
- comparaison des métriques principales,
- visualisations pour justifier le **choix du modèle final**.


## 1. Imports et configuration

Ce notebook a pour objectif de comparer plusieurs modèles de classification du churn dans un cadre méthodologique rigoureux. Afin de garantir la cohérence avec l’API et le dashboard, l’analyse repose sur les 10 variables retenues dans le pipeline final. Le dataset étant déséquilibré, l’évaluation ne repose pas uniquement sur le ROC-AUC : la PR-AUC, le recall et le F1-score sont également pris en compte.

```python

from pathlib import Path
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.exceptions import ConvergenceWarning, UndefinedMetricWarning
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    fbeta_score,
    make_scorer,
    precision_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)
from sklearn.model_selection import (
    StratifiedKFold,
    cross_val_predict,
    cross_validate,
    train_test_split,
)
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

warnings.filterwarnings("ignore", category=ConvergenceWarning)
warnings.filterwarnings("ignore", category=UndefinedMetricWarning)

sns.set_theme(style="whitegrid")

RANDOM_STATE = 42
TEST_SIZE = 0.2
CV_SPLITS = 5
TARGET = "churn"

FEATURES = [
    "gender",
    "age",
    "tenure_months",
    "contract_type",
    "monthly_logins",
    "weekly_active_days",
    "avg_session_time",
    "monthly_fee",
    "payment_failures",
    "nps_score",
]

# Le notebook peut être ouvert depuis la racine du projet ou depuis le dossier Code/
if (Path.cwd() / "SOURCE" / "dataset.csv").exists():
    DATA_PATH = Path.cwd() / "SOURCE" / "dataset.csv"
elif (Path.cwd().parent / "SOURCE" / "dataset.csv").exists():
    DATA_PATH = Path.cwd().parent / "SOURCE" / "dataset.csv"
else:
    raise FileNotFoundError("Impossible de trouver SOURCE/dataset.csv")
```

## 2. Chargement des données

```python

df = pd.read_csv(DATA_PATH)

X = df[FEATURES].copy()
y = df[TARGET].astype(int).copy()

print(f"Dataset : {df.shape[0]} lignes, {df.shape[1]} colonnes")
print(f"Taux de churn : {y.mean():.2%}")

display(df[FEATURES + [TARGET]].head())
```

Le dataset utilisé contient 10 000 clients. La variable cible `churn` représente environ 10,21 % des observations, ce qui confirme un déséquilibre de classes modéré mais réel. Ce point est important, car un modèle pourrait obtenir des performances apparemment correctes tout en détectant mal les churners. Cela justifie l’utilisation de métriques adaptées comme la PR-AUC et le recall.

## 3. Préparation du pipeline

```python

numeric_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features),
])

def build_pipeline(model):
    return Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", model),
    ])

def choose_best_threshold(y_true, y_proba):
    thresholds = np.arange(0.10, 0.91, 0.05)
    best_threshold = 0.50
    best_f2 = -1

    for threshold in thresholds:
        y_pred = (y_proba >= threshold).astype(int)
        f2 = fbeta_score(y_true, y_pred, beta=2, zero_division=0)

        if f2 > best_f2:
            best_f2 = f2
            best_threshold = round(float(threshold), 2)

    return best_threshold
```

Le prétraitement est intégré dans un pipeline `scikit-learn` combinant imputation, standardisation des variables numériques et encodage one-hot des variables catégorielles. Cette approche limite le risque de data leakage, car toutes les transformations sont apprises uniquement sur les données d’entraînement avant d’être appliquées au jeu de test.

## 4. Split train / test et définition des modèles

```python

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    stratify=y,
    random_state=RANDOM_STATE,
)

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=RANDOM_STATE,
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=-1,
    ),
    "Gradient Boosting": GradientBoostingClassifier(
        random_state=RANDOM_STATE,
    ),
    "MLP": MLPClassifier(
        hidden_layer_sizes=(64, 32),
        max_iter=500,
        early_stopping=True,
        random_state=RANDOM_STATE,
    ),
}

cv = StratifiedKFold(n_splits=CV_SPLITS, shuffle=True, random_state=RANDOM_STATE)

scoring = {
    "roc_auc": "roc_auc",
    "pr_auc": "average_precision",
    "precision": make_scorer(precision_score, zero_division=0),
    "recall": make_scorer(recall_score, zero_division=0),
    "f1": make_scorer(f1_score, zero_division=0),
}

print(f"Train : {X_train.shape} | Test : {X_test.shape}")
```

## 5. Entraînement, cross-validation et optimisation du seuil

```python

results = []
pipelines = {}
test_predictions = {}
test_probabilities = {}
best_thresholds = {}

for model_name, model in models.items():
    pipeline = build_pipeline(model)

    cv_results = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
    )

    oof_proba = cross_val_predict(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        method="predict_proba",
        n_jobs=-1,
    )[:, 1]

    best_threshold = choose_best_threshold(y_train, oof_proba)
    best_thresholds[model_name] = best_threshold

    pipeline.fit(X_train, y_train)

    y_proba_test = pipeline.predict_proba(X_test)[:, 1]
    y_pred_test = (y_proba_test >= best_threshold).astype(int)

    results.append({
        "model": model_name,
        "cv_roc_auc_mean": np.mean(cv_results["test_roc_auc"]),
        "cv_pr_auc_mean": np.mean(cv_results["test_pr_auc"]),
        "cv_precision_mean": np.mean(cv_results["test_precision"]),
        "cv_recall_mean": np.mean(cv_results["test_recall"]),
        "cv_f1_mean": np.mean(cv_results["test_f1"]),
        "threshold": best_threshold,
        "test_roc_auc": roc_auc_score(y_test, y_proba_test),
        "test_pr_auc": average_precision_score(y_test, y_proba_test),
        "test_precision": precision_score(y_test, y_pred_test, zero_division=0),
        "test_recall": recall_score(y_test, y_pred_test, zero_division=0),
        "test_f1": f1_score(y_test, y_pred_test, zero_division=0),
    })

    pipelines[model_name] = pipeline
    test_predictions[model_name] = y_pred_test
    test_probabilities[model_name] = y_proba_test

results_df = pd.DataFrame(results).sort_values(
    by=["cv_pr_auc_mean", "cv_recall_mean", "cv_roc_auc_mean"],
    ascending=False,
).reset_index(drop=True)

display(results_df.round(4))
```

La comparaison des modèles repose sur une validation croisée stratifiée à 5 folds, ce qui rend les résultats plus robustes qu’un simple split unique. En complément, un seuil de décision est optimisé à partir des probabilités issues de la validation croisée en maximisant le F2-score. Ce choix donne davantage d’importance au recall, ce qui est pertinent dans un problème de churn où manquer un client à risque peut avoir un coût business important.

## 6. Lecture des résultats

Le modèle final est choisi selon la logique suivante :

1. **PR-AUC moyenne en cross-validation**
2. **Recall moyen en cross-validation**
3. **ROC-AUC moyenne en cross-validation**

Ce choix est cohérent avec un problème de churn, où la classe positive est minoritaire et où l'on cherche à **détecter un maximum de churners**.


```python

best_model_name = results_df.loc[0, "model"]
best_pipeline = pipelines[best_model_name]
best_threshold = best_thresholds[best_model_name]

print(f"Modèle retenu : {best_model_name}")
print(f"Seuil retenu : {best_threshold}")
```

Les résultats montrent que les modèles d’ensemble obtiennent les meilleures performances globales. Le Random Forest présente la meilleure PR-AUC moyenne en validation croisée (≈ 0,251), devant Gradient Boosting (≈ 0,243), Logistic Regression (≈ 0,192) et MLP (≈ 0,078). 

Sur le jeu de test, Gradient Boosting obtient le meilleur F1-score (≈ 0,377), tandis que Random Forest reste très compétitif avec un meilleur compromis en validation croisée. Logistic Regression atteint un recall élevé, mais au prix d’une précision faible, ce qui traduit un grand nombre de faux positifs. Le MLP, quant à lui, est nettement moins performant et n’apparaît pas comme un bon candidat final sur ce dataset.

## 7. Comparaison visuelle des modèles

```python

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

plot_df = results_df.sort_values("cv_pr_auc_mean", ascending=True)
axes[0].barh(plot_df["model"], plot_df["cv_pr_auc_mean"])
axes[0].set_title("PR-AUC moyenne en cross-validation")
axes[0].set_xlabel("Score")

plot_df = results_df.sort_values("test_f1", ascending=True)
axes[1].barh(plot_df["model"], plot_df["test_f1"])
axes[1].set_title("F1-score sur le jeu de test")
axes[1].set_xlabel("Score")

plt.tight_layout()
plt.show()
```

Les résultats montrent que les modèles d’ensemble obtiennent les meilleures performances globales. Le Random Forest présente la meilleure PR-AUC moyenne en validation croisée (≈ 0,251), devant Gradient Boosting (≈ 0,243), Logistic Regression (≈ 0,192) et MLP (≈ 0,078). 

Sur le jeu de test, Gradient Boosting obtient le meilleur F1-score (≈ 0,377), tandis que Random Forest reste très compétitif avec un meilleur compromis en validation croisée. Logistic Regression atteint un recall élevé, mais au prix d’une précision faible, ce qui traduit un grand nombre de faux positifs. Le MLP, quant à lui, est nettement moins performant et n’apparaît pas comme un bon candidat final sur ce dataset.

## 8. Courbes ROC sur le jeu de test

```python

plt.figure(figsize=(8, 6))

for model_name, y_proba in test_probabilities.items():
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    auc_score = roc_auc_score(y_test, y_proba)
    plt.plot(fpr, tpr, label=f"{model_name} (AUC={auc_score:.3f})")

plt.plot([0, 1], [0, 1], linestyle="--", color="gray")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Courbes ROC comparatives")
plt.legend()
plt.show()
```

Les courbes ROC montrent que Random Forest et Gradient Boosting dominent la comparaison en termes de capacité de discrimination globale. Toutefois, dans un contexte déséquilibré, le ROC-AUC ne suffit pas à lui seul pour choisir le modèle final. C’est pourquoi il est interprété conjointement avec la PR-AUC et les métriques de classification à seuil optimisé.

## 9. Matrices de confusion

```python

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for ax, model_name in zip(axes, models.keys()):
    ConfusionMatrixDisplay.from_predictions(
        y_test,
        test_predictions[model_name],
        ax=ax,
        colorbar=False,
    )
    ax.set_title(f"{model_name}\n(seuil={best_thresholds[model_name]})")

plt.tight_layout()
plt.show()
```

L’analyse des matrices de confusion permet de mieux comprendre les compromis entre les modèles. Logistic Regression détecte davantage de churners, mais génère aussi beaucoup de faux positifs. À l’inverse, les modèles d’ensemble offrent un compromis plus équilibré entre sensibilité et précision. Le choix du seuil optimisé, souvent inférieur à 0,5, montre qu’il est nécessaire d’abaisser le seuil standard pour mieux capter les clients à risque dans un contexte de churn.

## 10. Importance des variables

```python

classifier = best_pipeline.named_steps["classifier"]
preprocessor_fitted = best_pipeline.named_steps["preprocessor"]

if hasattr(preprocessor_fitted, "get_feature_names_out"):
    feature_names = preprocessor_fitted.get_feature_names_out()
else:
    feature_names = None

importance_df = None

if feature_names is not None and hasattr(classifier, "feature_importances_"):
    importance_df = pd.DataFrame({
        "feature": feature_names,
        "importance": classifier.feature_importances_,
    }).sort_values("importance", ascending=False)

elif feature_names is not None and hasattr(classifier, "coef_"):
    importance_df = pd.DataFrame({
        "feature": feature_names,
        "importance": np.abs(classifier.coef_[0]),
    }).sort_values("importance", ascending=False)

if importance_df is None:
    print("Le modèle final ne fournit pas d'importance native.")
else:
    display(importance_df.head(10))

    plt.figure(figsize=(10, 5))
    sns.barplot(
        data=importance_df.head(10),
        x="importance",
        y="feature",
        orient="h",
    )
    plt.title(f"Top 10 des variables importantes — {best_model_name}")
    plt.xlabel("Importance")
    plt.ylabel("Variable")
    plt.tight_layout()
    plt.show()
```

L’interprétation du modèle final montre que l’ancienneté (`tenure_months`) et l’activité (`monthly_logins`) jouent un rôle central dans la prédiction. Viennent ensuite le temps moyen de session, le score NPS, l’âge et les incidents de paiement. À l’inverse, le genre et le type de contrat semblent moins discriminants dans la configuration actuelle. Cette hiérarchie est cohérente avec une logique métier où la fidélité, l’usage du service et les signaux de friction de paiement constituent des indicateurs plus informatifs que les caractéristiques démographiques simples.

## 11. Conclusion

```python

summary_cols = [
    "model",
    "cv_pr_auc_mean",
    "cv_recall_mean",
    "cv_roc_auc_mean",
    "threshold",
    "test_pr_auc",
    "test_recall",
    "test_f1",
]
display(results_df[summary_cols].round(4))
```

En conclusion, l’analyse comparative montre que les modèles classiques de type ensemble sont les plus adaptés à ce problème. Le Random Forest est retenu comme modèle final pour sa robustesse en validation croisée et son bon compromis global. Le Gradient Boosting constitue une alternative crédible, tandis que Logistic Regression reste intéressante comme baseline interprétable. Le MLP est conservé pour répondre à l’exigence de comparaison avec un modèle de deep learning, mais ses performances restent insuffisantes dans ce contexte.
